```python

# ==================================================
# ANEXO B - PUNTO 2
# ==================================================

import requests
import pandas as pd
from pathlib import Path

# 1. Definir fuentes oficiales
fuentes = pd.DataFrame([
    {
        "Fuente": "NASA FIRMS",
        "Producto": "VIIRS S-NPP 2024",
        "URL": "https://firms.modaps.eosdis.nasa.gov/download/",
        "Metodo": "Descarga histórica del ZIP oficial"
    },
    {
        "Fuente": "NASA POWER",
        "Producto": "Datos meteorológicos diarios",
        "URL": "https://power.larc.nasa.gov/",
        "Metodo": "API REST"
    },
    {
        "Fuente": "Copernicus",
        "Producto": "Sentinel-2 MSI L2A",
        "URL": "https://dataspace.copernicus.eu/browser/",
        "Metodo": "Exportación multiespectral GeoTIFF"
    }
])

print("FUENTES OFICIALES")
display(fuentes)

# 2. Consulta reproducible a NASA POWER
url = "https://power.larc.nasa.gov/api/temporal/daily/point"

parametros = {
    "latitude": 3.85,
    "longitude": -71.55,
    "start": "20240109",
    "end": "20240115",
    "community": "AG",
    "parameters": "T2M,RH2M,PRECTOTCORR",
    "format": "JSON",
    "time-standard": "UTC"
}

respuesta = requests.get(url, params=parametros, timeout=60)
respuesta.raise_for_status()

datos = respuesta.json()["properties"]["parameter"]

print("\nURL COMPLETA DE CONSULTA:")
print(respuesta.url)

print("\nVARIABLES RECUPERADAS:")
for variable, serie in datos.items():
    print(variable, ":", len(serie), "días")

# 3. Guardar evidencia
Path("evidencias_anexo_b").mkdir(exist_ok=True)

fuentes.to_csv(
    "evidencias_anexo_b/fuentes_oficiales.csv",
    index=False
)

with open(
    "evidencias_anexo_b/consulta_nasa_power.txt",
    "w",
    encoding="utf-8"
) as archivo:
    archivo.write(respuesta.url + "\n")

print("\nEvidencia de consulta guardada.")
```

```python

# ==================================================
# ANEXO B - VERIFICACIÓN DE USO DE KAGGLE
# Examina las celdas de código del cuaderno .ipynb
# ==================================================

import json
from google.colab import files

print("Selecciona el cuaderno .ipynb de tu proyecto")
subidos = files.upload()

nombre = next(
    (n for n in subidos if n.endswith(".ipynb")),
    None
)

if nombre is None:
    raise ValueError("Debes cargar un archivo .ipynb")

with open(nombre, "r", encoding="utf-8") as f:
    cuaderno = json.load(f)

patrones = ["kaggle", "kagglehub", "/kaggle/input"]
coincidencias = []

for numero, celda in enumerate(cuaderno["cells"], start=1):
    if celda.get("cell_type") != "code":
        continue

    codigo = "".join(celda.get("source", []))

    for linea_numero, linea in enumerate(
        codigo.splitlines(), start=1
    ):
        if any(p in linea.lower() for p in patrones):
            coincidencias.append({
                "celda": numero,
                "linea": linea_numero,
                "contenido": linea.strip()
            })

print("\nRESULTADO DE LA VERIFICACIÓN")
print("Cuaderno revisado:", nombre)
print("Coincidencias encontradas:", len(coincidencias))

if coincidencias:
    for c in coincidencias:
        print(c)
else:
    print("No se encontraron referencias a Kaggle en las celdas de código.")

# Guardar resultado
with open(
    "evidencias_anexo_b/verificacion_kaggle.txt",
    "w",
    encoding="utf-8"
) as f:
    f.write(f"Cuaderno: {nombre}\n")
    f.write(f"Coincidencias: {len(coincidencias)}\n")
    for c in coincidencias:
        f.write(str(c) + "\n")
```
